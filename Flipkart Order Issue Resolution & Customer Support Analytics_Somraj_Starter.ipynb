## Milestone 3 · Prepare reliable order records
**From the team:** Operations has noticed repeated IDs, inconsistent order labels, and delivery-date issues. Without cleaning, the team may miscount orders or report misleading delays.

**Your responsibility:** Produce a clean order-level source. Keep one record per order ID, normalize order statuses, ensure order amounts are numeric, and make order and delivery dates usable for comparisons. Treat non-delivered orders with no actual delivery date differently from invalid/missing dates on delivered orders. Explain your handling of ambiguous values.

**Handover evidence:** Create **`orders_clean`**, preserving key source fields: `order_id`, `customer_id`, `customer_city`, `order_date`, `promised_delivery_date`, `actual_delivery_date`, `product_category`, `order_amount`, `order_status`. Include a concise cleaning note.

```python
# MILESTONE 3: CLEAN AND STANDARDIZE ORDER RECORDS

# 1. Create a separate cleaned version of the orders dataset
#    while preserving the original data.

# 2. Identify and remove duplicate order records based on
#    order_id, and standardize inconsistent order statuses.

# 3. Ensure order amounts and date-related columns have
#    appropriate data formats for further analysis.

# 4. Investigate missing delivery information, distinguish
#    valid missing dates from data quality issues, and
#    summarize the cleaning results.


# Write your implementation below.




# Create a copy of the original orders dataset
orders_clean = orders_df.copy()

# Check duplicate order records
print("Duplicate order records:")
print(orders_clean.duplicated(subset="order_id").sum())

# Remove duplicate orders
orders_clean = orders_clean.drop_duplicates(subset="order_id", keep="first")

# Standardize order status values
orders_clean["order_status"] = (
    orders_clean["order_status"].str.strip().str.title()
)

# Convert date columns into datetime format
orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"], errors="coerce"
)

orders_clean["promised_delivery_date"] = pd.to_datetime(
    orders_clean["promised_delivery_date"], errors="coerce"
)

orders_clean["actual_delivery_date"] = pd.to_datetime(
    orders_clean["actual_delivery_date"], errors="coerce"
)

# Convert order amount into numeric format
orders_clean["order_amount"] = pd.to_numeric(
    orders_clean["order_amount"], errors="coerce"
)

# Check missing delivery dates
print("\nMissing actual delivery dates:")
print(orders_clean["actual_delivery_date"].isnull().sum())

# Check missing delivery dates according to order status
print("\nMissing delivery dates by order status:")
display(
    orders_clean.groupby("order_status")["actual_delivery_date"]
    .apply(lambda x: x.isnull().sum())
)

# Find delivered orders with missing actual delivery dates
incomplete_delivery = orders_clean[
    (orders_clean["order_status"] == "Delivered") &
    (orders_clean["actual_delivery_date"].isnull())
]

print("\nDelivered orders with missing delivery information:")
display(incomplete_delivery)

# Check missing values after cleaning
print("\nMissing values after cleaning:")
display(orders_clean.isnull().sum())

# Display cleaned dataset information
print("\nCleaned dataset information:")
orders_clean.info()

# Display cleaned dataset
print("\nFirst 5 cleaned records:")
display(orders_clean.head())

# Check for remaining duplicate orders
print("\nRemaining duplicate order IDs:")
print(orders_clean["order_id"].duplicated().sum())

print("\nTotal orders after cleaning:")
print(len(orders_clean))
```

## Milestone 4 · Make issue records consistent
**From the team:** Refunds and customer complaints were compiled by different teams. A single order can have multiple genuine issue records, so removing all repeated order IDs would lose useful evidence.

**Your responsibility:** Create dependable refund and complaint data. Resolve duplicate refund/complaint IDs, standardize status labels, make refund amounts and all event dates usable, and surface records with missing order links. Preserve legitimate multiple issues against the same order.

**Handover evidence:** Create **`refunds_clean`** and **`complaints_clean`**, retaining the original business fields and consistent status values. Show a small before/after quality summary.

```python
# MILESTONE 4: CLEAN REFUND AND COMPLAINT DATA

# 1. Create cleaned versions of the refund and complaint
#    datasets without modifying the original records.

# 2. Identify and handle duplicate refund and complaint
#    records using their respective unique identifiers.

# 3. Standardize refund and complaint status values,
#    monetary amounts, and date formats.

# 4. Identify missing order references and other data
#    quality issues, then display the cleaned datasets.


# Write your implementation below.



# Create copies of both datasets
refunds_clean = refunds_df.copy()
complaints_clean = complaints_df.copy()


# Clean refund data

# Check and remove duplicate refund records
print("Duplicate refund records:")
print(refunds_clean["refund_id"].duplicated().sum())

refunds_clean = refunds_clean.drop_duplicates(
    subset="refund_id", keep="first"
)

# Standardize refund status
refunds_clean["refund_status"] = (
    refunds_clean["refund_status"].str.strip().str.title()
)

# Convert refund amount to numeric
refunds_clean["refund_amount"] = pd.to_numeric(
    refunds_clean["refund_amount"], errors="coerce"
)

# Convert refund date to datetime
refunds_clean["refund_date"] = pd.to_datetime(
    refunds_clean["refund_date"], errors="coerce"
)

# Check missing order references
missing_refund_orders = refunds_clean[
    ~refunds_clean["order_id"].isin(orders_clean["order_id"])
]

print("\nRefund records with invalid order IDs:")
display(missing_refund_orders)


# Clean complaint data

# Check and remove duplicate complaint records
print("\nDuplicate complaint records:")
print(complaints_clean["complaint_id"].duplicated().sum())

complaints_clean = complaints_clean.drop_duplicates(
    subset="complaint_id", keep="first"
)

# Standardize complaint status and type
complaints_clean["complaint_status"] = (
    complaints_clean["complaint_status"].str.strip().str.title()
)

complaints_clean["complaint_type"] = (
    complaints_clean["complaint_type"].str.strip().str.title()
)

# Convert complaint date to datetime
complaints_clean["complaint_date"] = pd.to_datetime(
    complaints_clean["complaint_date"], errors="coerce"
)

# Check missing order references
missing_complaint_orders = complaints_clean[
    ~complaints_clean["order_id"].isin(orders_clean["order_id"])
]

print("\nComplaint records with invalid order IDs:")
display(missing_complaint_orders)


# Check missing values in both cleaned datasets
print("\nMissing values in refunds:")
display(refunds_clean.isnull().sum())

print("\nMissing values in complaints:")
display(complaints_clean.isnull().sum())


# Display cleaned datasets
print("\nCleaned refund records:")
display(refunds_clean.head())

print("\nCleaned complaint records:")
display(complaints_clean.head())

# Display final dataset information
print("\nRefund dataset information:")
refunds_clean.info()

print("\nComplaint dataset information:")
complaints_clean.info()
```

## Milestone 5 · Measure delivery reliability
**From the team:** The delivery lead needs a factual picture of late deliveries and where they occur before assigning support capacity.

**Your responsibility:** Analyze order volumes by status and revenue by product category. For completed deliveries with valid promised and actual dates, measure lateness relative to the promised day. Identify delayed orders and compare delayed-order counts across cities. Do not classify an undelivered shipment as an on-time delivery simply because its actual date is blank.

**Handover evidence:** Add numeric **`delivery_delay_days`** to **`orders_clean`** (positive = days late, zero = on date, negative = early; missing where not computable). Create **`city_delay_summary`** containing `customer_city` and `delayed_order_count`. Display the top five cities.

```python
# MILESTONE 5: ANALYZE DELIVERY PERFORMANCE

# 1. Analyze the distribution of order statuses and calculate
#    the total order amount across product categories.

# 2. Compare promised and actual delivery dates to determine
#    delivery delays for successfully delivered orders.

# 3. Identify orders delivered after their promised dates
#    and analyze delivery delays across customer cities.

# 4. Display the five cities with the highest number of
#    delayed deliveries and identify delivered orders
#    with incomplete delivery information.


# Write your implementation below.



# Check the number of orders in each status
print("Order status distribution:")
display(orders_clean["order_status"].value_counts())


# Calculate total revenue for each product category
print("\nRevenue by product category:")
category_revenue = orders_clean.groupby("product_category")["order_amount"].sum().reset_index()

category_revenue = category_revenue.sort_values(
    by="order_amount", ascending=False
)

display(category_revenue)


# Calculate delivery delay in days
orders_clean["delivery_delay_days"] = pd.NA

# Consider only delivered orders with both delivery dates available
delivered_orders = (
    (orders_clean["order_status"] == "Delivered") &
    (orders_clean["promised_delivery_date"].notna()) &
    (orders_clean["actual_delivery_date"].notna())
)

orders_clean.loc[delivered_orders, "delivery_delay_days"] = (
    orders_clean.loc[delivered_orders, "actual_delivery_date"] -
    orders_clean.loc[delivered_orders, "promised_delivery_date"]
).dt.days

# Convert delay column to numeric
orders_clean["delivery_delay_days"] = pd.to_numeric(
    orders_clean["delivery_delay_days"], errors="coerce"
)


# Display delivery delay information
print("\nDelivery delay summary:")
display(orders_clean["delivery_delay_days"].describe())


# Identify orders delivered late
delayed_orders = orders_clean[
    orders_clean["delivery_delay_days"] > 0
]

print("\nOrders delivered late:")
display(delayed_orders.head())

print("\nTotal delayed orders:")
print(len(delayed_orders))


# Count delayed orders across cities
city_delay_summary = (
    delayed_orders.groupby("customer_city")
    .size()
    .reset_index(name="delayed_order_count")
)

city_delay_summary = city_delay_summary.sort_values(
    by="delayed_order_count", ascending=False
)

print("\nTop 5 cities with the highest delayed deliveries:")
display(city_delay_summary.head(5))


# Find delivered orders with missing delivery information
incomplete_deliveries = orders_clean[
    (orders_clean["order_status"] == "Delivered") &
    (
        orders_clean["promised_delivery_date"].isna() |
        orders_clean["actual_delivery_date"].isna()
    )
]

print("\nDelivered orders with incomplete delivery information:")
display(incomplete_deliveries)

print("\nTotal delivered orders with incomplete information:")
print(len(incomplete_deliveries))
```

## Milestone 6 · Connect refunds to original orders
**From the team:** Finance can investigate refund requests, but needs the corresponding order context. Some orders have no refund request; those orders must remain in the overview.

**Your responsibility:** Investigate refund outcomes, including requests pending or failed. Summarize the refund evidence *at order level* before joining to orders, keeping all original orders in the result. Contrast requested refund amounts and identify high-value cases.

**Handover evidence:** Create **`refunds_by_order`** with one row per `order_id` and the fields `refund_count`, `total_refund_amount`, `pending_or_failed_refund_count`. Create **`orders_refunds_df`** with one row per order. Show refund-status and high-value case summaries.

```python
# MILESTONE 6: CONNECT REFUNDS WITH CUSTOMER ORDERS

# 1. Summarize refund records at the order level, including
#    the number of refund requests and total requested amount.

# 2. Identify pending and failed refund requests and calculate
#    the number of unresolved refunds associated with each order.

# 3. Combine refund information with cleaned order records,
#    ensuring orders without refund requests are retained.

# 4. Analyze refund amounts by status, identify the five orders
#    with the highest requested refund amounts, and display
#    the combined order-refund dataset.


# Write your implementation below.

# Check the different refund statuses
print("Refund status summary:")
display(refunds_clean["refund_status"].value_counts())


# Count pending and failed refunds
refunds_clean["pending_or_failed"] = refunds_clean["refund_status"].isin(
    ["Pending", "Failed"]
)


# Create a summary of refunds for each order
refunds_by_order = refunds_clean.groupby("order_id").agg(
    refund_count=("refund_id", "count"),
    total_refund_amount=("refund_amount", "sum"),
    pending_or_failed_refund_count=("pending_or_failed", "sum")
).reset_index()


# Join refund information with the orders dataset
orders_refunds_df = orders_clean.merge(
    refunds_by_order,
    on="order_id",
    how="left"
)


# Replace missing refund values with zero
refund_columns = [
    "refund_count",
    "total_refund_amount",
    "pending_or_failed_refund_count"
]

orders_refunds_df[refund_columns] = (
    orders_refunds_df[refund_columns].fillna(0)
)

# Convert refund counts to integers
orders_refunds_df["refund_count"] = (
    orders_refunds_df["refund_count"].astype(int)
)

orders_refunds_df["pending_or_failed_refund_count"] = (
    orders_refunds_df["pending_or_failed_refund_count"].astype(int)
)


# Analyze refund amounts by status
print("\nTotal refund amount by status:")
refund_status_summary = refunds_clean.groupby("refund_status").agg(
    refund_count=("refund_id", "count"),
    total_refund_amount=("refund_amount", "sum")
).reset_index()

display(refund_status_summary)


# Find the five orders with the highest refund amounts
print("\nTop 5 orders by requested refund amount:")

top_refund_orders = refunds_by_order.sort_values(
    by="total_refund_amount", ascending=False
).head(5)

display(top_refund_orders)


# Display the order-level refund summary
print("\nRefund summary by order:")
display(refunds_by_order.head())


# Display the combined dataset
print("\nOrders with refund information:")
display(orders_refunds_df.head())

print("\nCombined dataset shape:")
print(orders_refunds_df.shape)

print("\nTotal orders in combined dataset:")
print(orders_refunds_df["order_id"].nunique())
```

```python
# MILESTONE 7: ANALYZE CUSTOMER COMPLAINTS

# 1. Analyze customer complaints based on complaint type
#    and resolution status.

# 2. Calculate the total number of complaints associated
#    with each order and identify orders with repeated complaints.

# 3. Identify customers who have raised multiple complaints
#    and rank them based on complaint frequency.

# 4. Highlight the five most common complaint categories
#    and summarize unresolved customer complaints.


# Write your implementation below.



# Check complaints by type
print("Complaints by category:")
display(complaints_clean["complaint_type"].value_counts())


# Check complaints by resolution status
print("\nComplaints by status:")
display(complaints_clean["complaint_status"].value_counts())


# Count complaints for each order
complaints_by_order = complaints_clean.groupby("order_id").agg(
    complaint_count=("complaint_id", "count")
).reset_index()

# Identify orders with multiple complaints
repeated_complaints = complaints_by_order[
    complaints_by_order["complaint_count"] > 1
]

print("\nOrders with multiple complaints:")
display(repeated_complaints.sort_values(
    by="complaint_count", ascending=False
))

print("\nTotal orders with multiple complaints:")
print(len(repeated_complaints))


# Count complaints raised by each customer
customer_complaints = complaints_clean.groupby("customer_id").agg(
    complaint_count=("complaint_id", "count")
).reset_index()

customer_complaints = customer_complaints.sort_values(
    by="complaint_count", ascending=False
)

print("\nCustomers with the highest number of complaints:")
display(customer_complaints.head(10))


# Find the five most common complaint categories
print("\nTop 5 complaint categories:")
top_complaints = complaints_clean["complaint_type"].value_counts().head(5)
display(top_complaints)


# Identify unresolved complaints
unresolved_complaints = complaints_clean[
    complaints_clean["complaint_status"].isin(["Open", "In Progress"])
]

print("\nUnresolved customer complaints:")
display(unresolved_complaints.head(10))

print("\nTotal unresolved complaints:")
print(len(unresolved_complaints))


# Count open and in-progress complaints for each order
open_complaints_by_order = complaints_clean[
    complaints_clean["complaint_status"].isin(["Open", "In Progress"])
].groupby("order_id").agg(
    open_complaint_count=("complaint_id", "count")
).reset_index()

print("\nOrders with unresolved complaints:")
display(open_complaints_by_order.head())
```

## Milestone 8 · Deliver one order investigation view
**From the team:** Support agents should no longer need to manually look up the three source files for every case. The lead wants a dependable one-row-per-order working table.

**Your responsibility:** Combine cleaned order details with the aggregated refund and complaint information. Retain orders with no associated issues, represent absent issue counts as zero, and verify no orders were accidentally duplicated or lost when joining. Note any issue records referring to unknown orders.

**Handover evidence:** Create **`final_orders_df`** with one row per order, including `order_id`, `customer_id`, `customer_city`, `order_status`, `delivery_delay_days`, `refund_count`, `total_refund_amount`, `pending_or_failed_refund_count`, `complaint_count`, `open_complaint_count`. Show a row-count reconciliation.

```python
# MILESTONE 8: CREATE A CONSOLIDATED ORDER ISSUE DATASET

# 1. Prepare order-level summaries of refund requests,
#    unresolved refunds, total refund amounts, customer
#    complaints, and open complaints.

# 2. Combine the cleaned orders dataset with the refund
#    and complaint summaries using the common order_id.

# 3. Ensure all original orders are retained and handle
#    missing issue information appropriately.

# 4. Validate that the consolidated dataset contains one
#    record per order without unintended duplicate rows,
#    and display the final combined information.


# Write your implementation below.



# Create refund summary for each order
refund_summary = refunds_clean.groupby("order_id").agg(
    refund_count=("refund_id", "count"),
    total_refund_amount=("refund_amount", "sum"),
    pending_or_failed_refund_count=("pending_or_failed", "sum")
).reset_index()


# Create complaint summary for each order
complaint_summary = complaints_clean.groupby("order_id").agg(
    complaint_count=("complaint_id", "count")
).reset_index()


# Count unresolved complaints for each order
open_complaint_summary = complaints_clean[
    complaints_clean["complaint_status"].isin(["Open", "In Progress"])
].groupby("order_id").agg(
    open_complaint_count=("complaint_id", "count")
).reset_index()


# Combine all information with the orders dataset
final_orders_df = orders_clean.merge(
    refund_summary, on="order_id", how="left"
)

final_orders_df = final_orders_df.merge(
    complaint_summary, on="order_id", how="left"
)

final_orders_df = final_orders_df.merge(
    open_complaint_summary, on="order_id", how="left"
)


# Fill missing issue counts with zero
issue_columns = [
    "refund_count",
    "total_refund_amount",
    "pending_or_failed_refund_count",
    "complaint_count",
    "open_complaint_count"
]

final_orders_df[issue_columns] = final_orders_df[issue_columns].fillna(0)

# Convert count columns to integers
count_columns = [
    "refund_count",
    "pending_or_failed_refund_count",
    "complaint_count",
    "open_complaint_count"
]

final_orders_df[count_columns] = final_orders_df[count_columns].astype(int)


# Check the consolidated dataset
print("First 5 records:")
display(final_orders_df.head())

print("\nDataset shape:")
print(final_orders_df.shape)

print("\nUnique orders:")
print(final_orders_df["order_id"].nunique())

print("\nDuplicate order IDs:")
print(final_orders_df["order_id"].duplicated().sum())

print("\nMissing values:")
display(final_orders_df.isnull().sum())
```

Assign `High` when at least two signals apply, `Medium` for exactly one, and `Low` otherwise. Surface incomplete delivery-date records separately rather than assuming they were on time.

```python
# MILESTONE 9: IDENTIFY HIGH-PRIORITY SUPPORT CASES

# 1. Identify orders delivered more than three days late,
#    orders with pending or failed refunds, and orders
#    associated with multiple customer complaints.

# 2. Evaluate how many of these issue conditions are
#    satisfied by each order.

# 3. Classify orders as High priority when two or more
#    conditions are satisfied, Medium when exactly one
#    is satisfied, and Low when none are satisfied.

# 4. Summarize the priority distribution, display the
#    top ten High-priority orders, and flag incomplete
#    delivery information for separate review.


# Write your implementation below.



# Identify orders delivered more than 3 days late
final_orders_df["delayed_over_3_days"] = (
    (final_orders_df["order_status"] == "Delivered") &
    (final_orders_df["delivery_delay_days"] > 3)
).fillna(False).astype(bool)


# Identify orders with pending or failed refunds
final_orders_df["refund_needs_attention"] = (
    final_orders_df["pending_or_failed_refund_count"] > 0
).fillna(False).astype(bool)


# Identify orders with more than one complaint
final_orders_df["repeated_complaints"] = (
    final_orders_df["complaint_count"] > 1
).fillna(False).astype(bool)


# Count the number of issue conditions for each order
final_orders_df["issue_count"] = (
    final_orders_df["delayed_over_3_days"].astype(int) +
    final_orders_df["refund_needs_attention"].astype(int) +
    final_orders_df["repeated_complaints"].astype(int)
)


# Assign support priority
def get_priority(count):
    if count >= 2:
        return "High"
    elif count == 1:
        return "Medium"
    else:
        return "Low"


final_orders_df["support_priority"] = (
    final_orders_df["issue_count"].apply(get_priority)
)


# Flag incomplete delivery information
final_orders_df["incomplete_delivery_information"] = (
    (final_orders_df["order_status"] == "Delivered") &
    (
        final_orders_df["actual_delivery_date"].isna() |
        final_orders_df["promised_delivery_date"].isna()
    )
)


# Display priority distribution
print("Support priority distribution:")
display(final_orders_df["support_priority"].value_counts())


# Create priority summary
priority_summary = final_orders_df.groupby("support_priority").agg(
    order_count=("order_id", "count")
).reset_index()

display(priority_summary)


# Display the top 10 high-priority orders
print("\nTop 10 high-priority orders:")
high_priority_orders = final_orders_df[
    final_orders_df["support_priority"] == "High"
].sort_values(by="issue_count", ascending=False)

display(high_priority_orders.head(10))


# Display incomplete delivery records
print("\nIncomplete delivery information:")
display(
    final_orders_df[
        final_orders_df["incomplete_delivery_information"]
    ]
)
```
